```python
import svgwrite

# -------------------------------------------------
# SVG SETUP
# -------------------------------------------------
WIDTH, HEIGHT = 900, 900

ACTIVE_STROKE_WIDTH = 1.6
INACTIVE_STROKE_WIDTH = 0.0


dwg = svgwrite.Drawing(
    filename="lanes.svg",
    size=(f"{WIDTH}px", f"{HEIGHT}px")
)

# White background
dwg.add(dwg.rect(insert=(0, 0), size=(WIDTH, HEIGHT), fill="white"))

# -------------------------------------------------
# ARROW GEOMETRY (24x24 ICONS)
# -------------------------------------------------
def left_turn_arrow(group, active=True):
    group.add(dwg.path(
        d=(
            "M9.04297 3.54303c0.39052 -0.39052 1.02353 -0.39052 "
            "1.41403 0 0.3905 0.39053 0.3905 1.02356 0 1.41406 "
            "L7.41406 8.00006H16c1.6568 0.00004 3 1.34317 "
            "3 3.00004v8c0 0.5522 -0.4478 0.9999 -1 1 "
            "-0.5523 0 -1 -0.4478 -1 -1v-8c0 -0.5523 "
            "-0.4478 -1 -1 -1H7.41406L10.457 13.043 "
            "c0.3905 0.3906 0.3905 1.0236 0 1.4141 "
            "s-1.0235 0.3904 -1.41403 0l-4.75 -4.75001 "
            "c-0.39052 -0.39052 -0.39052 -1.02354 0 -1.41406z",
        ),
        fill="black" if active else "#b0b0b0",
        stroke="black",
        stroke_width=ACTIVE_STROKE_WIDTH if active else INACTIVE_STROKE_WIDTH
    ))


def right_turn_arrow(group, active=True):
    group.add(dwg.path(
        d=(
            "M13.543 3.54303c0.3905 -0.39052 1.0235 -0.39052 1.414 0"
            "l4.75 4.75c0.3905 0.39053 0.3905 1.02355 0 1.41406"
            "l-4.75 4.75001c-0.3905 0.3905 -1.0235 0.3905 -1.414 0"
            "s-0.3905 -1.0235 0 -1.4141l3.0429 -3.0429H8"
            "c-0.55228 0 -1 0.4477 -1 1v8"
            "c-0.00004 0.5522 -0.44774 1 -1 1"
            "s-0.99996 -0.4478 -1 -1v-8"
            "c0 -1.65689 1.34315 -3.00004 3 -3.00004"
            "h8.5859L13.543 4.95709"
            "c-0.3905 -0.39051 -0.3905 -1.02353 0 -1.41406"
        ),
        fill="black" if active else "#b0b0b0",
        stroke="black",
        stroke_width=ACTIVE_STROKE_WIDTH if active else INACTIVE_STROKE_WIDTH
    ))


def straight_arrow(group, active=True):
    group.add(dwg.path(
        d=(
            "M12.0001 2c0.5523 0.00004 1 0.44774 1 1v15.0859"
            "l4.043 -4.0429c0.3905 -0.3905 1.0235 -0.3905 1.4141 0"
            "c0.3904 0.3905 0.3904 1.0235 0 1.414"
            "l-5.75 5.75"
            "c-0.0466 0.0466 -0.0984 0.0874 -0.1534 0.1241"
            "c-0.0391 0.0261 -0.0794 0.0493 -0.1211 0.0693"
            "c-0.0455 0.0219 -0.0926 0.0408 -0.1416 0.0557"
            "c-0.0167 0.005 -0.0338 0.0085 -0.0507 0.0127"
            "c-0.0156 0.0038 -0.031 0.0086 -0.0469 0.0117"
            "c-0.0071 0.0014 -0.0144 0.0017 -0.0215 0.0029"
            "c-0.0559 0.0097 -0.1132 0.0166 -0.1719 0.0166"
            "c-0.0566 0 -0.1119 -0.0066 -0.166 -0.0156"
            "c-0.0104 -0.0018 -0.0209 -0.0028 -0.0312 -0.0049"
            "c-0.0115 -0.0023 -0.0229 -0.0051 -0.0342 -0.0078"
            "c-0.0206 -0.0049 -0.0413 -0.0094 -0.0615 -0.0156"
            "c-0.0487 -0.0149 -0.0955 -0.0339 -0.1407 -0.0557"
            "c-0.0398 -0.0192 -0.0787 -0.0407 -0.1162 -0.0654"
            "c-0.0566 -0.0374 -0.1094 -0.0802 -0.1572 -0.128"
            "l-5.75001 -5.75"
            "c-0.39052 -0.3905 -0.39052 -1.0235 0 -1.414"
            "c0.39053 -0.3905 1.02355 -0.3905 1.41406 0"
            "l4.04295 4.0429V3c0 -0.55228 0.4477 -1 1 -1"
        ),
        fill="black" if active else "#b0b0b0",
        stroke="black",
        stroke_width=ACTIVE_STROKE_WIDTH if active else INACTIVE_STROKE_WIDTH
    ))

# -------------------------------------------------
# LOCAL DRAW HELPERS (NO ROTATION HERE)
# -------------------------------------------------
def draw_left(g, x, y, active=True):
    gg = dwg.g(transform=f"translate({x},{y}) translate(-12,-12)")
    left_turn_arrow(gg, active)
    g.add(gg)


def draw_straight(g, x, y, active=True):
    gg = dwg.g(transform=f"translate({x},{y}) rotate(180) translate(-12,-12)")
    straight_arrow(gg, active)
    g.add(gg)


def draw_right(g, x, y, active=True):
    gg = dwg.g(transform=f"translate({x},{y}) translate(-12,-12)")
    right_turn_arrow(gg, active)
    g.add(gg)

# -------------------------------------------------
# UTIL: NORMALIZE LANES
# -------------------------------------------------
def normalize_lanes(lanes):
    out = {}
    for k, v in lanes.items():
        if isinstance(v, int):
            out[k] = [True] * v
        else:
            out[k] = list(v)
    return out

# -------------------------------------------------
# MAIN FUNCTION
# -------------------------------------------------
def draw_lane_arrows(
    x,
    y,
    angle=0,
    scale=3.5,
    lanes=None,
    spacing=26,
    divider_height=100,
    divider_offset=24
):
    if lanes is None:
        lanes = {"left": 1, "straight": 1, "right": 1}

    lanes = normalize_lanes(lanes)

    g = dwg.g(
        transform=f"translate({x},{y}) rotate({angle}) scale({scale})"
    )

    # divider line (left of leftmost arrow)
    g.add(dwg.line(
        start=(-divider_offset, -divider_height / 2),
        end=(-divider_offset, divider_height / 2),
        stroke="black",
        stroke_width=2
    ))

    cx = 0

    for lane_type in ("left", "straight", "right"):
        for active in lanes.get(lane_type, []):
            if lane_type == "left":
                draw_left(g, cx, 0, active)
            elif lane_type == "straight":
                draw_straight(g, cx, 0, active)
            elif lane_type == "right":
                draw_right(g, cx, 0, active)

            cx += spacing

    dwg.add(g)

# -------------------------------------------------
# EXAMPLES (YOUR CASE)
# -------------------------------------------------

```

```python
def draw_intersection_lanes(
    all_movements,
    active_movements,
    label=None,
    center_x=450,
    center_y=450,
    offset=200,
    label_offset=18
):
    """
    all_movements  : list of all lanes that exist
    active_movements : list of lanes that are active
    """

    ANGLE_BY_DIR = {"N": 0, "E": 90, "S": 180, "W": 270}

    POSITION_BY_DIR = {
        "N": (center_x, center_y + offset),
        "S": (center_x, center_y - offset),
        "E": (center_x - offset, center_y),
        "W": (center_x + offset, center_y),
    }

    # Parse lane structure
    approaches = parse_movements(all_movements, active_movements)

    # Title
    if label:
        draw_title(label)

    # Draw each approach
    for direction, lanes in approaches.items():
        angle = ANGLE_BY_DIR[direction]
        x, y = POSITION_BY_DIR[direction]

        draw_lane_arrows(
            x=x,
            y=y,
            angle=angle,
            lanes=lanes
        )


def draw_title(text, y=60, font_size=80):
    dwg.add(
        dwg.text(
            text,
            insert=(WIDTH / 2, y),
            text_anchor="middle",
            font_size=f"{font_size}px",
            font_weight="bold",
            fill="black"
        )
    )

def parse_movements(all_movements, active_movements):
    """
    Returns:
    {
        "N": {
            "left":     [True, False],
            "straight": [True, True, False],
            "right":    [False]
        },
        ...
    }
    """

    active_set = set(active_movements)
    approaches = {}

    for mv in all_movements:
        direction = mv[0]   # N / S / E / W
        movement  = mv[2]   # L / T / R

        if direction not in approaches:
            approaches[direction] = {
                "left": [],
                "straight": [],
                "right": []
            }

        is_active = mv in active_set

        if movement == "L":
            approaches[direction]["left"].append(is_active)
        elif movement == "T":
            approaches[direction]["straight"].append(is_active)
        elif movement == "R":
            approaches[direction]["right"].append(is_active)
        else:
            raise ValueError(f"Unknown movement: {mv}")

    return approaches
```

```python
converted_phases = [
    {
        "nema_phase_combo": "2+5",
        "movements": ["EBT", "SBR", "EBL"]
    },
    {
        "nema_phase_combo": "2+6",
        "movements": ["EBT", "WBT"]
    },
    {
        "nema_phase_combo": "1+5",
        "movements": ["WBL", "NBR", "SBR", "EBL"]
    },
    {
        "nema_phase_combo": "1+6",
        "movements": ["WBL", "NBR", "WBT"]
    },
    {
        "nema_phase_combo": "3+7",
        "movements": ["NBL", "EBR", "SBL", "WBR"]
    },
    {
        "nema_phase_combo": "3+8",
        "movements": ["NBL", "EBR", "NBT"]
    },
    {
        "nema_phase_combo": "4+7",
        "movements": ["SBT", "SBL", "WBR"]
    },
    {
        "nema_phase_combo": "4+8",
        "movements": ["SBT", "NBT"]
    }
]
```

```python
all_movements=[
        "NBL", "NBT", "NBR",
        "SBL", "SBT", "SBR",
        "EBL", "EBT", "EBT", "EBR",
        "WBL", "WBT", "WBT", "WBR",
    ]


for phase in converted_phases:
    
    
    # start a fresh SVG for each phase
    dwg = svgwrite.Drawing(
        filename=f"phase_{phase['nema_phase_combo']}.svg",
        size=(f"{WIDTH}px", f"{HEIGHT}px")
    )

    # background
    dwg.add(dwg.rect(insert=(0, 0), size=(WIDTH, HEIGHT), fill="white"))

    # title
    draw_title(f"NEMA Phase {phase['nema_phase_combo']}")

    # draw intersection
    draw_intersection_lanes(
        all_movements=all_movements,
        active_movements=phase["movements"],
        label=None   # title already drawn
    )

    # save
    dwg.save()
```
